```python
# 1. Write a function check_even_odd(n) that accepts a number and prints whether the number is even or odd. 

def check_even_odd(num):
    if num>0:
        if num%2==0:
            return "Prime"
        else:
            return "Not a Prime"
    else:
        return 'Enter a positive number'
        
check_even_odd(0)
```

```python
# 4. Write a function find_largest(a, b, c) that finds the largest among three numbers.

def find_largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c


print(find_largest(16, 25, 65))
```

```python
# 5. Write a function sum_natural(n) that calculates the sum of numbers from 1 to n.

def sum_natural(n):
    
    total = 0
    
    for i in range (0, n+1):
        total = total + i
        
    return total

sum_natural(6)
```

```python
# 7. Write a function factorial(n) that calculates the factorial of a number.

def factorial(n=int(input())):
    
    total = 1
    
    for i in range (1, n+1):
        
        total *= i
        
    return total

factorial()
```

```python
# 8. Write a function count_digits(n) that counts the number of digits in a number.

n = int(input("Enter a number: "))

def count_digits(n):
    n = abs(n)
    count = 0

    if n == 0:
        return 1

    while n > 0:
        count += 1
        n //= 10

    return count


count_digits(n)
```

```python
# 9. Write a function reverse_number(n) that reverses a given number.

n = int(input("Enter a number: "))

def reverse_number(n):
    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n //= 10

    return reverse

reverse_number(n)
```

```python
# 10. Write a function check_prime(n) that checks whether a number is prime. 

def prime(num):
    
    if num <=1:
        return "Not a Prime Number"

    for n in range(2,int(num**0.5)+1):
            if num % n == 0:
                return "Not a Prime Number"
            
    return "Prime"
                
prime(89)
```

```python
# 12. Write a function count_vowels(text) that counts the number of vowels in a string.

def count_vowels(text):
    count = 0

    for char in text:
        if char.lower() in "aeiou":
            count += 1

    return count


print(count_vowels("I code python"))
```

```python
# 13. Write a function count_consonants(text) that counts the number of consonants in a string.

def count_consonants(text):
    count = 0

    for char in text:
        if char.isalpha() and char.lower() not in "aeiou":
            count += 1

    return count


print(count_consonants("Glorious Purpose"))
```

```python
# 14. Write a function count_vowels_consonants(text) that separately counts vowels and consonants.

def count_vowels_consonants(text):
    vowels = 0
    consonants = 0

    for char in text:
        if char.isalpha():
            if char.lower() in "aeiou":
                vowels += 1
            else:
                consonants += 1

    return vowels, consonants


print(count_vowels_consonants("Excellent"))
```

```python
# 15. Write a function reverse_string(text) that reverses a string.

def reverse_string(text):
    reverse = ""

    for char in text:
        reverse = char + reverse

    return reverse


print(reverse_string("Zomato"))
```

```python
# 16. Write a function check_palindrome(text) that checks whether a string is a palindrome.

def check_palindrome(text):
    reverse = ""

    for char in text:
        reverse = char + reverse

    return text == reverse


print(check_palindrome("madam"))
print(check_palindrome("mister"))
```

```python
# 18. Write a function character_frequency(text, ch) that counts how many times a particular character occurs.

def character_frequency(text, ch):
    count = 0

    for char in text:
        if char == ch:
            count += 1

    return count


print(character_frequency("brightest nights", "t"))
```

```python
# 19. Write a function remove_spaces(text) that removes all spaces from a string.

def remove_spaces(text):
    result = ""

    for char in text:
        if char != " ":
            result += char

    return result


print(remove_spaces("Hi nice to meet you"))
```

```python
# 21. Write a function count_case(text) that counts:
#  uppercase characters
#  lowercase characters
#  digits
#  spaces

def count_case(text):
    uppercase = 0
    lowercase = 0
    digits = 0
    spaces = 0

    for char in text:
        if char.isupper():
            uppercase += 1
        elif char.islower():
            lowercase += 1
        elif char.isdigit():
            digits += 1
        elif char == " ":
            spaces += 1

    return uppercase, lowercase, digits, spaces


print(count_case("it is Easy to code in python 2234"))
```

```python
# 27. Write a function find_longest_word(text) that finds the longest word in a sentence.

def find_longest_word(text):
    words = text.split()
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


print(find_longest_word("Python is very powerful"))
```

```python
# 28. Write a function that separately count
# a
# e
# i
# o
# u

def count_each_vowel(text):
    a = 0
    e = 0
    i = 0
    o = 0
    u = 0

    for char in text.lower():
        if char == "a":
            a += 1
        elif char == "e":
            e += 1
        elif char == "i":
            i += 1
        elif char == "o":
            o += 1
        elif char == "u":
            u += 1

    return a, e, i, o, u


print(count_each_vowel("Peaky Blinders Orders"))
```

```python
# 29. Write a function star_pattern(n) that prints:
# *
# **
# ***
# ****
# *****
# For n = 5.

def star_pattern(n):
    for i in range(1, n + 1):
        print("*" * i)


star_pattern(5)
```

```python
# 30. Write a function number_pattern(n) that prints:
# 1
# 12
# 123
# 1234
# 12345

def number_pattern(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print(j, end="")
        print()


number_pattern(5)
```
